```python
import numpy as np 
import pandas as pd 
import difflib
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
data = pd.read_csv("movies.csv")
```

```python
data.shape
```

```python
data.isnull().sum()
```

```python
selected_features = ['genres', 'keywords', 'tagline', 'cast', 'director']
selected_features
```

```python

for features in selected_features:
    data[features] = data[features].fillna('')
```

```python
combined_features = data['genres']+' '+data['keywords']+' '+data['tagline']+' '+data['cast']+' '+data['director']
```

```python
combined_features
```

```python
combined_features.head()
```

```python
vectorizer = TfidfVectorizer()
```

```python
feature_vectors= vectorizer.fit_transform(combined_features)
```

```python
feature_vectors
```

```python
# cosine Similarity

similarity = cosine_similarity(feature_vectors)
```

```python
similarity.shape
```

```python
## getting the user name from user

movie_name= input("Enter the name of the movie : ")
```

```python
list_of_all_movie = data['title'].tolist()
```

```python
list_of_all_movie
```

```python
find_close_match = difflib.get_close_matches(movie_name, list_of_all_movie)

find_close_match
```

```python
close_match = find_close_match[0]
```

```python
index_of_movie = data[data.title == close_match]['index'].values[0]
index_of_movie
```

```python
similarity_score = list(enumerate(similarity[index_of_movie]))
similarity_score
```

```python
sorted_similiar_movies = sorted(similarity_score, key= lambda x:x[1], reverse=True)
sorted_similiar_movies
```

```python
print("Movie Suggested for You : \n")

i = 1
for movie in sorted_similiar_movies:
    index = movie[0]
    title_from_index = data[data.index == index]['title'].values[0]
    if (i<30):
        print(i, '.', title_from_index)
        i+=1
```
