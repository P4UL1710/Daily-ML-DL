```python
import numpy as np
import pandas as pd
from sklearn.model_selection import StratifiedShuffleSplit
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
data = pd.read_csv("mail_data.csv")
```

```python
data.head()
```

```python
data.info()
```

```python
data['Category'].value_counts()
```

```python
data.isnull().sum()
```

Label Encoding

```python
data.loc[data['Category'] == 'spam', 'Catagory',] = 0
data.loc[data['Category'] == 'ham', 'Catagory',] = 0
```

Spam -> 0

Ham-> 1

```python
## Separating the label and target 

X = data['Message']

Y = data['Category']
```

```python
X.shape
```

```python
Y.shape
```

```python
Y
```

```python
data.replace({'Category': {'spam':0,'ham':1}}, inplace=True)
```

```python
split = StratifiedShuffleSplit()

for train_index, test_index in split.split(X, Y):
    X_train = X.iloc[train_index]
    X_test = X.iloc[test_index]
    Y_train = Y.iloc[train_index]
    Y_test = Y.iloc[test_index]
```

```python
X_train
```

```python
## Feature Extraction

Vectorizer = TfidfVectorizer(min_df=1 , stop_words='english', lowercase=True)

X_train_features = Vectorizer.fit_transform(X_train)
X_test_features = Vectorizer.transform(X_test)
```

```python
X_train_features
```

```python
model = LogisticRegression()
```

```python
model.fit(X_train_features, Y_train)
```

```python
## Prediction and evalutaion

Train_pred = model.predict(X_train_features)

Train_score = accuracy_score(Train_pred, Y_train)

Train_score
```

```python
Test_pred = model.predict(X_test_features)

Test_score = accuracy_score(Test_pred, Y_test)

Test_score
```

```python
# Building a predictive system

input_data = input("Give the message you want to confirm is real or not: ")

# Convert input into list
input_data_features = Vectorizer.transform([input_data])

prediction = model.predict(input_data_features)

if prediction[0] == 0:
    print("Model has predicted it as a spam email")
else:
    print("Model has predicted it as a ham email")
```
