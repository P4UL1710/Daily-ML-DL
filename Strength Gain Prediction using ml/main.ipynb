```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv('supplement_impact_data.csv')
```

```python
data.shape
```

```python
data.isnull().sum()
```

```python
data.head()
```

```python
data['Gender'].value_counts()
```

```python
data.replace({'Gender': {'Non-Binary': 0, 'Male': 1, 'Female': 2}}, inplace=True)
```

```python
data['Supplement'].value_counts()
```

```python
data.replace({'Supplement': {'Both': 0, 'Mass Gainer': 1, 'Creatine Monohydrate': 2}}, inplace=True)
```

```python
data["Strength_Gain"] = data["Strength_Gain"].str.replace("%", "").astype(int)
```

```python
data.info()
```

```python
data['Primary_Benefit'].value_counts()
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

data['Primary_Benefit'] = le.fit_transform(data['Primary_Benefit'])
```

```python
data.info()
```

```python
X = data.drop(columns=['ID', 'Strength_Gain'], axis=1)
Y = data['Strength_Gain']
```

```python
from sklearn.model_selection import train_test_split

X_train , X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
X_train.shape
```

```python
Y_train.shape
```

```python
from xgboost import XGBRegressor

model = XGBRegressor()
```

```python
model.fit(X_train, Y_train)
```

```python
from sklearn import metrics

Train_pred = model.predict(X_train)

Train_score = metrics.r2_score(Y_train, Train_pred)

Train_score
```

```python
Test_pred = model.predict(X_test)

Test_score = metrics.r2_score(Y_test, Test_pred)

Test_score
```

```python
# Building a predictive sysem 
input_data = [64,1,2,5,79.1,81.2,3]

input_data_as_np_array = np.asarray(input_data)

input_data_reshaped = input_data_as_np_array.reshape(1,-1)

Prediction = model.predict(input_data_reshaped)

print(f"The model has predicted the Strength gained is :{Prediction[0]:.2f}%")
```
